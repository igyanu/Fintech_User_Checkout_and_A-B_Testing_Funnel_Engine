```python
import numpy as np
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest, proportion_effectsize
from statsmodels.stats.power import NormalIndPower
```

```python
df = pd.read_csv(r'C:\Users\hp\Downloads\fintech/A_B_Testing.csv')
df
```

```python
# splitting data for variant 1 and variant 2

variant_1 = df[df.Variant=='Variant_1'][['Variant','Converted']]
variant_2 = df[df.Variant=='Variant_2'][['Variant','Converted']]
```

```python
# seeing conversion rate in both cases

size_1 = variant_1.shape[0]
conversion_1 = variant_1.Converted.sum()

size_2 = variant_2.shape[0]
conversion_2 = variant_2.Converted.sum()

conversion_rate_1 = conversion_1 / size_1
conversion_rate_2 = conversion_2 / size_2

print(f'Conversion Rate in Variant 1: {conversion_rate_1}')
print(f'Conversion Rate in Variant 2: {conversion_rate_2}')
```

```python
# calculating statistical power

effect_size = proportion_effectsize(conversion_rate_1, conversion_rate_2)
power = NormalIndPower().solve_power(
    effect_size = effect_size,
    nobs1 = size_1,
    alpha = .05,
    ratio = size_1/size_2
)
print(f'Power is {power}\n')
print(f'There is {power*100}% probability that test will detect the actual change if it exists')
```

Performing prorpotional Z test to check whether 1% change in conversion rate is random or due to payment gateway changes.

level of significance = 1%   
H0: Conversion rate in both variants don't differs significantly.   
H1: Conversion rate in variant 2 is significantly greater than in variant 1.

```python
# proportions z test

conversion = [conversion_1, conversion_2]
size = [size_1, size_2]
z_stat, p_value = proportions_ztest(conversion, size)
print(f'Calculated p value is {round(p_value,3)}\n')
if(p_value < 0.05):
    print(f'Now, we have sttistical evidence to prove uplift in conversion is due to UI change.\nReject Null Hypothesis')
```
